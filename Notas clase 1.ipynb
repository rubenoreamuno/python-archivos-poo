```python
# Listas []
# Se pueden iterar --> recorrer

lista = [1,2,3,4,5]
```

```python
lista[2] = 4
```

```python
lista
```

```python
for i in lista:
  print(i)
```

```python
# Tuplas ()
# Inmutables --> no se pueden alterar
tupla = (('nombre', 'Juan'), ('apellido', 'Perez'))
```

```python
tupla[1][0]
```

```python
tupla[0][1] = "Ruben"
```

```python
for i in tupla:
    print(i)
```

```python
# Diccionarios {}
# 

orden = {
    'nombre': 'Juan',
    'apellido': 'Perez',
    'edad': 20,
    'direccion': 'Calle 123',
    'telefono': '1234567890',
    'email': 'juan.perez@gmail.com',
    'ciudad': 'Bogota',
    'pais': 'Colombia',
    'codigo_postal': '12345',
    'productos' : ['producto1', 'producto2', 'producto3']
}
```

```python
orden.items()
```
